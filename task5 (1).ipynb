```python
list=[12,13,14,15,16,19,21,25,28,31,45,49,52,55,59,62,68,80,83,86]
print(list)
even_list=[]
odd_list=[]


for  num in list:
   if num % 2 == 0:
     even_list.append(num)

   else:
       odd_list.append(num)

       print("even number list",even_list)
       print("odd number list",odd_list)
       print("max number list",max(list))
       print("min number list",min(list))
       print("total number even",len(even_list))
       print("total number odd",len(odd_list))
```
